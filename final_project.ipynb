```python
# ==========================================================
# CELL 2 → IMPORT LIBRARIES
# ==========================================================

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, HTML
import ipywidgets as widgets
```

```python
# ==========================================================
# CELL 3 → TITLE & INTRODUCTION
# ==========================================================

title_html = """
<div style="
    background: linear-gradient(90deg,#1B5E20,#1565C0);
    padding:20px;
    border-radius:15px;
    color:white;
    text-align:center;
    margin-bottom:20px;
">
    <h1>🌱 Smart Sustainable Finance Planner</h1>
    <h3>Middle Class Sustainable Finance Dashboard</h3>
    <p>
    ✅ Budget Planning &nbsp;&nbsp;
    ✅ Savings & SIP &nbsp;&nbsp;
    ✅ Emergency Fund &nbsp;&nbsp;
    ✅ Government Schemes
    </p>
</div>
"""

display(HTML(title_html))
```

```python
# ==========================================================
# CELL 4 → USER INPUTS
# ==========================================================

name = input("Enter Your Name: ")

monthly_salary = float(input("Monthly Salary (₹): "))
age = int(input("Your Age: "))
family_members = int(input("Family Members: "))

print("\n========== Monthly Expenses ==========\n")

rent = float(input("House Rent / EMI: "))
groceries = float(input("Groceries: "))
transport = float(input("Transport: "))
school_fees = float(input("School Fees: "))
electricity = float(input("Electricity + Internet: "))
entertainment = float(input("Entertainment: "))
loan_emi = float(input("Loan EMI: "))
other_expenses = float(input("Other Expenses: "))

print("\n========== Savings & Insurance ==========\n")

emergency_fund = float(input("Emergency Fund Saved: "))
investments = float(input("Current Investments: "))

health_insurance = input(
    "Health Insurance (No Insurance / Employer Insurance Only / Basic Insurance / Good Insurance): "
)
```

```python
# ==========================================================
# CELL 5 → FINANCIAL CALCULATIONS
# ==========================================================

total_expenses = (
    rent + groceries + transport +
    school_fees + electricity +
    entertainment + loan_emi +
    other_expenses
)

surplus = monthly_salary - total_expenses

savings_rate = (
    surplus / monthly_salary * 100
    if monthly_salary > 0 else 0
)

target_emergency = total_expenses * 6
```

```python
# ==========================================================
# CELL 6 → FINANCIAL ANALYSIS
# ==========================================================

analysis_html = f"""
<div style="
    background:#0D1B2A;
    padding:20px;
    border-radius:15px;
    color:white;
    margin-top:20px;
">

<h2>📈 Financial Analysis</h2>

<table style="width:100%; font-size:18px;">
<tr>
<td><b>Monthly Salary</b></td>
<td>₹{monthly_salary:,.0f}</td>
</tr>

<tr>
<td><b>Total Expenses</b></td>
<td>₹{total_expenses:,.0f}</td>
</tr>

<tr>
<td><b>Total Savings</b></td>
<td>₹{surplus:,.0f}</td>
</tr>

<tr>
<td><b>Savings Rate</b></td>
<td>{savings_rate:.1f}%</td>
</tr>

</table>

</div>
"""

display(HTML(analysis_html))
```

```python
# ==========================================================
# CELL 7 → SMART INSIGHTS
# ==========================================================

print("\n========== SMART FINANCIAL INSIGHTS ==========\n")

if entertainment > 5000:
    print("⚠️ Entertainment expenses high hain.")

if groceries > 10000:
    print("⚠️ Grocery spending reduce kar sakte hain.")

if loan_emi > monthly_salary * 0.30:
    print("⚠️ EMI burden high hai.")

if surplus > 0:
    print(f"\n✅ Aap har month approx ₹{surplus:,.0f} save kar sakte hain.")
    print("\nSuggested Focus:")
    print("• SIP Investment")
    print("• PPF / NPS")
    print("• Emergency Fund")

else:
    print("❌ Expenses income se zyada hain.")
```

```python
# ==========================================================
# CELL 8 → EXPENSE DISTRIBUTION PIE CHART
# ==========================================================

labels = [
    "Rent",
    "Groceries",
    "Transport",
    "School",
    "Bills",
    "Entertainment",
    "Loan",
    "Other",
    "Savings"
]

values = [
    rent,
    groceries,
    transport,
    school_fees,
    electricity,
    entertainment,
    loan_emi,
    other_expenses,
    max(surplus, 0)
]

chart_df = pd.DataFrame({
    "Category": labels,
    "Amount": values
})

fig = px.pie(
    chart_df,
    names="Category",
    values="Amount",
    hole=0.45,
    title="📉 Expense Distribution"
)

fig.update_traces(
    textposition='inside',
    textinfo='percent+label'
)

fig.update_layout(
    height=700
)

fig.show()
```

```python
# ==========================================================
# CELL 9 → RECOMMENDED SAVINGS PLAN
# ==========================================================

if surplus > 0:

    savings_df = pd.DataFrame({

        "Investment Type": [
            "Emergency Fund",
            "Mutual Fund SIP",
            "PPF",
            "Gold Savings",
            "Extra Savings"
        ],

        "Monthly Amount (₹)": [
            int(surplus * 0.30),
            int(surplus * 0.30),
            int(surplus * 0.20),
            int(surplus * 0.10),
            int(surplus * 0.10)
        ]
    })

    print("\n========== RECOMMENDED SAVINGS PLAN ==========\n")

    display(savings_df)
```

```python
# ==========================================================
# CELL 10 → GOVERNMENT SAVING SCHEMES
# ==========================================================

schemes_df = pd.DataFrame({

    "Scheme": [
        "PPF",
        "NPS",
        "Atal Pension Yojana",
        "PMJJBY",
        "PMSBY"
    ],

    "Benefits": [
        "Long-term tax free savings",
        "Retirement pension benefits",
        "Guaranteed pension scheme",
        "₹2 lakh life insurance",
        "₹2 lakh accident insurance"
    ]
})

print("\n========== GOVERNMENT SAVING SCHEMES ==========\n")

display(schemes_df)
```

```python
# ==========================================================
# CELL 11 → EMERGENCY FUND STATUS
# ==========================================================

progress = min(
    emergency_fund / target_emergency,
    1.0
)

percentage = progress * 100

print("\n========== EMERGENCY FUND STATUS ==========\n")

print(f"Emergency Fund Saved: ₹{emergency_fund:,.0f}")
print(f"Target Emergency Fund: ₹{target_emergency:,.0f}")

print(f"\nCompletion Status: {percentage:.1f}%")
```

```python
# ==========================================================
# CELL 12 → FINAL ADVICE
# ==========================================================

print("\n========== FINAL ADVICE ==========\n")

advice = []

if entertainment > 5000:
    advice.append("⚠️ Entertainment expenses reduce kariye.")

if investments == 0:
    advice.append("✅ SIP investment start kariye.")

if emergency_fund < target_emergency:
    advice.append("✅ Emergency fund build kariye.")

if loan_emi > monthly_salary * 0.30:
    advice.append("⚠️ EMI burden kam kariye.")

for item in advice:
    print(item)

print(f"\n🎉 {name}, your sustainable finance report is ready!")
```
